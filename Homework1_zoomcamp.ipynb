```python
vehicles_asia = df.loc[
    df["origin"] == "Asia",
    ["vehicle_weight", "model_year"]
].head(7)

X = vehicles_asia.to_numpy()

print(X)
print(X.shape)
```

```python
Xt =X.T
print(Xt)
```

```python
XTX = Xt@X
XTX
```

```python
#Invert XTX
XTX_inv = np.linalg.inv(XTX)
print(XTX_inv)
```

```python
y= np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
type(y)
```

```python
w = XTX_inv@Xt@y
w
```
